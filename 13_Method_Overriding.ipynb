<font color="blue" size="6">Method Overriding in Python </font>

<p><b>Method overriding is an OOP concept in Python where a child class provides its own implementation of a method that is already defined in the parent class.
<p>In simple words, the child class changes the behavior of the parent class method.


```python
#Simple Bank Application Example
#Let's create a bank application with:
#•	Parent class: Bank
#•	Child class: SavingsAccount
#•	Child class: CurrentAccount
#Both child classes override the calculate_interest() method.
# Parent class
class Bank:
    def calculate_interest(self):
        print("Bank interest rate is 4%")

# Child class 1
class SavingsAccount(Bank):
    def calculate_interest(self):
        print("Savings Account interest rate is 7%")

# Child class 2
class CurrentAccount(Bank):
    def calculate_interest(self):
        print("Current Account interest rate is 2%")

# Creating objects
bank = Bank()
savings = SavingsAccount()
current = CurrentAccount()

# Calling methods
bank.calculate_interest()
savings.calculate_interest()
current.calculate_interest()

```

<p>Explanation:
<p>1.	The Bank class has a method called calculate_interest().
<p>2.	SavingsAccount inherits from Bank and overrides that method.
<p>3.	CurrentAccount also inherits from Bank and overrides the same method.
<p>4.	When we call the method using child-class objects, Python executes the child class's method instead of the parent class's method.


```python
#Using super() in Method Overriding
class Bank:
    def calculate_interest(self):
        print("Welcome to Bank")
        print("Base interest rate: 4%")
class SavingsAccount(Bank):
    def calculate_interest(self):
        super().calculate_interest()
        print("Savings interest rate: 7%")
account = SavingsAccount()
account.calculate_interest()
```
