```python
#1. Basic inheritance example
# Parent class
class Animal:
    def eat(self):
        print("Animal is eating")

# Child class
class Dog(Animal):
    def bark(self):
        print("Dog is barking")

# Creating object
d = Dog()

d.eat()   # Inherited method
d.bark()  # Child class method
```

<p>Here:
<p>Animal is the parent class.
<p>Dog is the child class.
<p>Dog inherits the eat() method from Animal.


```python
#Using super() in Python
#The super() function is used to call methods from the parent class.
```

```python
class Parent:
    def display(self):
        print("This is parent class")

class Child(Parent):
    def display(self):
        super().display()
        print("This is child class")

obj = Child()
obj.display()
```

```python
#Single Inheritance
#Meaning: One child class inherits from one parent class.
#Banking example: A SavingsAccount inherits the features of a BankAccount.
```

```python
# Parent class
class BankAccount:
    def deposit(self, amount):
        print("Deposited:", amount)

    def balance(self):
        print("Checking account balance")


# Child class
class SavingsAccount(BankAccount):
    def interest(self):
        print("Savings interest: 5%")


# Object creation
account = SavingsAccount()

account.deposit(1000)
account.balance()
account.interest()
```

<p>Explanation:
<p>BankAccount is the parent class.
<p>SavingsAccount is the child class.
<p>SavingsAccount inherits deposit() and balance().
<p>It also has its own interest() method.


```python
#Multiple Inheritance
#Meaning: One child class inherits from two or more parent classes.
#Banking example: A PremiumAccount inherits features from both BankServices and LoanServices.
```

```python
#Multiple Inheritance
# First parent class
class BankServices:
    def online_banking(self):
        print("Online banking available")


# Second parent class
class LoanServices:
    def loan(self):
        print("Loan facility available")


# Child class
class PremiumAccount(BankServices, LoanServices):
    def premium_benefits(self):
        print("Premium benefits available")


# Object creation
account = PremiumAccount()

account.online_banking()
account.loan()
account.premium_benefits()
```

<p>Explanation:
<p>BankServices is the first parent.
<p>LoanServices is the second parent.
<p>PremiumAccount inherits from both parents.
<p>The child can access methods from both parent classes.

<p>Multilevel Inheritance
<p>Meaning: A class inherits from another child class, creating multiple levels of inheritance.
<p>Banking example: BankAccount → SavingsAccount → StudentAccount.


```python
#Python program:
# Grandparent class
class BankAccount:
    def deposit(self):
        print("Money deposited")
# Parent class
class SavingsAccount(BankAccount):
    def interest(self):
        print("Interest rate: 5%")
# Child class
class StudentAccount(SavingsAccount):
    def student_benefits(self):
        print("Student account benefits")
# Object creation
account = StudentAccount()
account.deposit()
account.interest()
account.student_benefits()
```

<p>Explanation:
<p>BankAccount is the grandparent class.
<p>SavingsAccount inherits from BankAccount.
<p>StudentAccount inherits from SavingsAccount.
<p>StudentAccount can access methods from both previous levels.

```python
#Hierarchical Inheritance
#Meaning: Multiple child classes inherit from one parent class.
#Banking example: SavingsAccount, CurrentAccount, and FixedDeposit inherit from BankAccount.
```

```python
# Parent class
class BankAccount:
    def deposit(self):
        print("Money deposited")
# First child class
class SavingsAccount(BankAccount):
    def savings_interest(self):
        print("Savings interest: 5%")
# Second child class
class CurrentAccount(BankAccount):
    def overdraft(self):
        print("Overdraft facility available")
# Third child class
class FixedDeposit(BankAccount):
    def maturity(self):
        print("Fixed deposit maturity: 2 years")
# Objects
s = SavingsAccount()
c = CurrentAccount()
f = FixedDeposit()
s.deposit()
s.savings_interest()
c.deposit()
c.overdraft()
f.deposit()
f.maturity()
```

<p>Explanation:
<p>BankAccount is the common parent class.
<p>Three child classes inherit from it.
<p>Each child has its own specific functionality.
<p>All three children can access deposit().


```python
#Hybrid Inheritance
#Meaning: A combination of two or more types of inheritance.
#Banking example: A CustomerAccount inherits from both SavingsAccount and LoanAccount, while both inherit from BankAccount.
```

```python
# Parent class
class BankAccount:
    def account_details(self):
        print("Bank account details")
# First child
class SavingsAccount(BankAccount):
    def savings(self):
        print("Savings account available")


# Second child
class LoanAccount(BankAccount):
    def loan(self):
        print("Loan facility available")
# Multiple inheritance
class CustomerAccount(SavingsAccount, LoanAccount):
    def customer(self):
        print("Customer account created")
# Object creation
account = CustomerAccount()
account.account_details()
account.savings()
account.loan()
account.customer()
```

<p>Explanation:
<p>BankAccount is the parent class.
<p>SavingsAccount and LoanAccount inherit from BankAccount.
<p>CustomerAccount inherits from both child classes.
<p>This combines hierarchical and multiple inheritance.
