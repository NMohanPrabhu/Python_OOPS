```python
class Person:
  def __init__(self, name):
    self.name = name

  def greet(self):
    print("Hello, my name is " + self.name)

p1 = Person("Emil")
p1.greet() 
```

Methods with parameter

```python
class Calculator:
  def add(self, a, b):
    return a + b

  def multiply(self, a, b):
    return a * b

calc = Calculator()
print(calc.add(5, 3))
print(calc.multiply(4, 7)) 
```

Methods Accessing Properties
Methods can access and modify object properties using self:

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def get_info(self):
    return f"{self.name} is {self.age} years old"

p1 = Person("Tobias", 28)
print(p1.get_info()) 
```

Methods Modifying Properties:Methods can modify the properties of an object:

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def celebrate_birthday(self):
    self.age += 1
    print(f"Happy birthday! You are now {self.age}")
p1 = Person("Linus", 25)
p1.celebrate_birthday()
p1.celebrate_birthday() 
```

<p>__init__() — initializes the object's attributes when the object is created.
<p>self.age — stores the age belonging to that particular object.
<p>self.age += 1 — increases the stored age by one.
<p>p1.celebrate_birthday() — calls the method on the same object each time.

```python
#The __str__() Method
#The __str__() method is a special method that controls what is returned when the object is printed:
#Example: Without the __str__() method:
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
p1 = Person("Emil", 36)
print(p1)
```

```python
#Example: With the __str__() method:
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1) 
```

<p><b>Notice that __str__() uses return rather than print. It returns the string to the code that requested it.
<p><b>Python automatically calls p1.__str__() to obtain the object's readable string representation.
<p><b>What happens without __str__()?
<p><b>Python displays a default object representation, which is not very informative to a beginner.

Multiple Methods:A class can have multiple methods that work together:

```python
class Playlist:
  def __init__(self, name):
    self.name = name
    self.songs = []

  def add_song(self, song):
    self.songs.append(song)
    print(f"Added: {song}")

  def remove_song(self, song):
    if song in self.songs:
      self.songs.remove(song)
      print(f"Removed: {song}")

  def show_songs(self):
    print(f"Playlist '{self.name}':")
    for song in self.songs:
      print(f"- {song}")

my_playlist = Playlist("Favorites")
my_playlist.add_song("Bohemian Rhapsody")
my_playlist.add_song("Stairway to Heaven")
my_playlist.show_songs() 
```

```python
my_playlist.remove_song("Bohemian Rhapsody")
```

<p>Method call	Result
<p>my_playlist.add_song("Imagine")	Adds a song
<p>my_playlist.remove_song("Imagine")	Removes the specified song if it exists
<p>my_playlist.remove_song()	Raises a TypeError because the required argument is missing
<p>my_playlist.show_songs()	Displays all songs
