<font color="blue" size="6">Contructor Overloading</font>

<p>Constructor overloading means defining multiple constructors in the same class with different numbers or types of parameters.
<p>In languages such as Java and C++, we can define multiple constructors with different parameter lists. However, Python does not support traditional constructor overloading directly.
<p>If we define __init__() multiple times in a class, the last definition replaces the earlier ones.
<p>In Python, we commonly achieve similar behavior using default arguments, variable-length arguments (*args), or class methods.


<p>Example using default arguments
<p>Suppose we want to create a Student object in three ways:
<p>1.Without passing any values.
<p>2.By passing only the student's name.
<p>3.By passing both name and mark.


```python
class Student:
    def __init__(self, name="Unknown", mark=0):
        self.name = name
        self.mark = mark

    def display(self):
        print("Name:", self.name)
        print("Mark:", self.mark)


s1 = Student()
s2 = Student("Arun")
s3 = Student("Priya", 95)
s1.display()
s2.display()
s3.display()
```

<p><b>Explanation</b>
<p><b>s1 = Student()</b>
<p>No arguments are passed, so the default values are used: "Unknown" and 0.
<p><b>s2 = Student("Arun")</b>
<p>The name becomes "Arun". The mark remains 0.
<p><b>s3 = Student("Priya", 95)</b>
<p>Both arguments are passed, so the name is "Priya" and the mark is 95.
<p>The constructor uses default values to handle different numbers of arguments.


```python
#Example using *args
#The *args parameter allows a function to receive a variable number of positional arguments.
class Student:
    def __init__(self, *args):
        if len(args) == 0:
            self.name = "Unknown"
            self.mark = 0

        elif len(args) == 1:
            self.name = args[0]
            self.mark = 0

        elif len(args) == 2:
            self.name = args[0]
            self.mark = args[1]

        else:
            raise TypeError("Expected 0, 1, or 2 arguments")

    def display(self):
        print(self.name, self.mark)
s1 = Student()
s2 = Student("Arun")
s3 = Student("Priya", 95)
s1.display()
s2.display()
s3.display()
```

Here:
<p>len(args) == 0: no arguments.
<p>len(args) == 1: one argument.
<p>len(args) == 2: two arguments.
<p>Unlike the default-argument approach, *args lets us explicitly decide what to do based on the number of supplied arguments.


```python
#multiple __init__() methods
class Student:
    def __init__(self):
        print("No arguments")

    def __init__(self, name):
        print("Name:", name)


s1 = Student("Arun")
```

<p>The second __init__() replaces the first definition. Therefore, this works:
<p>s1 = Student("Arun")
<p>But this raises a TypeError:
<p>s1 = Student()
<p>Python uses the last defined __init__() method, which requires the name argument.
