<font color="Blue" size="6">Class Variables vs Instance Variables</font>

```python
#1. Instance Variables
#Instance variables belong to individual objects.
#Each object can have different values.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Student("Arun", 21)
s2 = Student("Bala", 22)
print(s1.name)
print(s2.name)
```

<p>Here, name and age are instance variables.
<p>•	s1 has its own name and age.
<p>•	s2 has its own name and age.


```python
#2. Class Variables
#Class variables are shared by all objects of a class.
#They are declared inside the class but outside the methods.
class Student:
    college = "ABC College"

    def __init__(self, name):
        self.name = name
s1 = Student("Arun")
s2 = Student("Bala")
print(s1.name, s1.college)
print(s2.name, s2.college)
```

<p>Here:
<p>•	college is a class variable.
<p>•	name is an instance variable.
<p>•	Both objects access the same class variable.


```python
#Changing Class Variables
class Employee:
    company = "Google"

e1 = Employee()
e2 = Employee()

Employee.company = "Microsoft"

print(e1.company)
print(e2.company)
```

<p>#Both objects see the updated class variable.
<p>#Important: If you assign a value using e1.company = "Amazon", it creates an instance variable for e1 and does not change the class variable.
