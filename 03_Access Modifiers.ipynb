<font color="blue" size="6"><B>Access Modifiers in Python</b></font>

<b>Public Members
<p>Public members can be accessed from anywhere.
<p>They can be accessed inside and outside the class.
<p>By default, all variables and methods in Python are public.


```python
class Student:
    def __init__(self):
        self.name = "John"   # Public variable

    def display(self):       # Public method
        print("Name:", self.name)

s1 = Student()

print(s1.name)
s1.display()
```

<b>Protected Members
<p>Protected members are indicated by a single underscore (_) before the variable or method name.
<p>They are intended to be accessed within the class and its subclasses.
<p>Python does not strictly prevent access from outside the class.


```python
class Student:
    def __init__(self):
        self._marks = 85

    def display(self):
        print("Marks:", self._marks)

s1 = Student()

print(s1._marks)
s1.display()
```

<b>Private Members
<p>Private members are indicated by two underscores (__) before the variable or method name.
<p>They are intended to be accessed only within the class.
<p>Python uses a mechanism called name mangling to make direct access more difficult.


```python
class Student:
    def __init__(self):
        self.__marks = 90

    def display(self):
        print("Marks:", self.__marks)

s1 = Student()

s1.display()

# This produces an AttributeError
print(s1.__marks)
#we will get attribute error since Here, __marks is a private variable.
```

```python
#Accessing a private variable using name mangling
class Student:
    def __init__(self):
        self.__marks = 90

s1 = Student()

print(s1._Student__marks)
```

#This demonstrates that Python private members are not strictly inaccessible. Name mangling mainly prevents accidental access.

```python
#Practical Example: Employee Class
class Employee:

    def __init__(self):
        self.name = "Ravi"       # Public
        self._salary = 30000     # Protected
        self.__bonus = 5000      # Private

    def display(self):
        print("Name:", self.name)
        print("Salary:", self._salary)
        print("Bonus:", self.__bonus)


e1 = Employee()

e1.display()

print(e1.name)
print(e1._salary)

# print(e1.__bonus)  # AttributeError
```
