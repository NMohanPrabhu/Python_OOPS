<Font color="blue" size="6">Magic Methods</font>

<p>Magic methods are special methods in Python that have double underscores (__) at the beginning and end of their names. They are also called dunder methods (Double UNDERscore).
<p>They allow you to define how objects behave with built-in Python operations.


```python
#Example 1: __init__() and __str__()
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name} is {self.age} years old"
s1 = Student("Arun", 21)
print(s1)
```

```python
#•  __init__() automatically runs when an object is created.
#•  __str__() automatically runs when you print an object.
```

```python
#Example 2: __add__()
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

a = Number(10)
b = Number(20)

print(a + b)
```

```python
#Here, __add__() defines what happens when we use the + operator between two objects.
```

```python
#Example 3: __len__()
class MyData:
    def __init__(self, data):
        self.data = data

    def __len__(self):
        return len(self.data)

obj = MyData([10, 20, 30, 40])

print(len(obj))
```

```python
#Magic methods are not usually called directly. Python automatically invokes them when you use corresponding operations.
```

```python
#Customize operator behavior in Python (Operator Overloading)
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks


s1 = Student(80)
s2 = Student(90)

print(s1 + s2)
```

<p>How does it work?
<p>When Python sees:
<p>s1 + s2
<p>It internally calls:
<p>s1.__add__(s2)
<p>Which executes: return self.marks + other.marks.So, instead of adding two objects directly, we customize + to add their marks.


```python
#Example 2: Customize the * operator
class Product:
    def __init__(self, price):
        self.price = price

    def __mul__(self, quantity):
        return self.price * quantity


p1 = Product(100)

print(p1 * 5)
```
