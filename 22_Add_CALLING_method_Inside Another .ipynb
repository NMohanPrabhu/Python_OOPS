<font color="blue" size="6">Calling a Method Inside Another Method </font>

<p>In Python, one method can call another method within the same class using the self keyword.
<p>This helps us reuse code instead of writing the same logic multiple times.


```python
class Calculator:

    def add(self):
        a = 10
        b = 20
        print("Addition:", a + b)

    def display(self):
        print("Welcome to Calculator")
        self.add()

obj = Calculator()
obj.display()
```

<b>How does it work?
<p>1.	obj.display() calls the display() method.
<p>2.	Python executes print("Welcome to Calculator").
<p>3.	self.add() calls the add() method.
<p>4.	The addition result is printed.


```python
#Example with Arguments and Return Values
#One method can pass arguments to another method and receive its return value.
class Calculator:

    def add(self, a, b):
        return a + b

    def multiply(self):
        result = self.add(10, 20)
        print("Multiplication:", result * 2)

obj = Calculator()
obj.multiply()
```

<b>Explanation:
<p>•	multiply() calls add(10, 20).
<p>•	add() returns 30.
<p>•	result stores 30.
<p>•	result * 2 gives 60.


```python
#3. Real-World Example
#Imagine a bank account. One method calculates the balance, and another displays the account details.
class Bank:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def get_balance(self):
        return self.balance
    def display(self):
        print("Account holder:", self.name)
        print("Balance:", self.get_balance())
b1 = Bank("Arun", 50000)
b1.display()
```

<p>Important Points
<p>•	self refers to the current object.
<p>•	Use self.method_name() to call another instance method.
<p>•	You can pass arguments to the called method.
<p>•	You can use the returned value from another method.
<p>•	Both methods should belong to the same object/class in this example.


```python
class Demo:

    def first(self):
        x = 10
        return x * 2

    def second(self):
        y = self.first()
        print(y + 5)

obj = Demo()
obj.second()
```
