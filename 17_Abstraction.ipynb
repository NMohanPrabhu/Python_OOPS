<font color="blue" size="6">Abstraction</font>

<p>Abstraction is an Object-Oriented Programming (OOP) concept that hides implementation details and shows only the essential features to the user.
<p>In simple words:
<p>Abstraction means hiding how something works and showing only what it does.


<p>Let's create a bank application using abstraction.
<p>We will create:
<p>•	Abstract class: BankAccount
<p>•	Child class: SavingsAccount
<p>•	Child class: CurrentAccount
<p>The abstract class defines what operations every bank account must support. The child classes define how those operations work.


```python
from abc import ABC, abstractmethod

# Abstract Class
class BankAccount(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def check_balance(self):
        pass


# Child Class 1
class SavingsAccount(BankAccount):

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance")

    def check_balance(self):
        print("Savings Balance:", self.balance)

# Child Class 2
class CurrentAccount(BankAccount):

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance")

    def check_balance(self):
        print("Current Balance:", self.balance)

# Creating objects
savings = SavingsAccount(10000)
current = CurrentAccount(20000)

# Calling methods
savings.deposit(5000)
savings.withdraw(2000)
savings.check_balance()

print("----------------")

current.deposit(10000)
current.withdraw(5000)
current.check_balance()
```

<p>Explanation:
<p>1.	ABC is imported from the abc module.
<p>2.	BankAccount(ABC) creates an abstract class.
<p>3.	@abstractmethod defines methods that child classes must implement.
<p>4.	pass indicates that the abstract methods have no implementation in the parent class.
<p>5.	SavingsAccount and CurrentAccount inherit from BankAccount.
<p>6.	Both child classes provide their own implementations.
<p>7.	We create objects of the child classes and call their methods.
