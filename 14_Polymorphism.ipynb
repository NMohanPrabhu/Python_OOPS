<font color="Blue" size="6">Polymorphism</font>

<p>Polymorphism means one name, many forms.
<p>In Python, polymorphism allows the same method name to perform different tasks depending on which object calls it.
<p>Let's understand polymorphism using a simple Bank Application.


<p><b><u>Real-world example</u></b>
<p><b>Imagine a bank has three types of accounts:</b>
<p>Savings Account,Current Account,Fixed Deposit Account
<p>All three accounts have a method called calculate_interest().But each account calculates interest differently.
<p>For example:
<p>Savings Account – 7% interest,Current Account – 2% interest,Fixed Deposit Account – 8% interest
<p>This is called polymorphism.


```python
# Parent class
class Bank:
    def calculate_interest(self):
        print("Bank interest calculation")

# Child class 1
class SavingsAccount(Bank):
    def calculate_interest(self):
        print("Savings Account Interest: 7%")

# Child class 2
class CurrentAccount(Bank):
    def calculate_interest(self):
        print("Current Account Interest: 2%")


# Child class 3
class FixedDepositAccount(Bank):
    def calculate_interest(self):
        print("Fixed Deposit Interest: 8%")

# Creating objects
savings = SavingsAccount()
current = CurrentAccount()
fixed = FixedDepositAccount()

# Polymorphism
accounts = [savings, current, fixed]

for account in accounts:
    account.calculate_interest()
```

<p>Explanation:
<p>1.	We create a parent class called Bank.
<p>2.	We create three child classes.
<p>3.	Each child class has the same method name: calculate_interest().
<p>4.	Each class provides a different implementation.
<p>5.	We store all objects in a single list.
<p>6.	Using a for loop, we call the same method on every object.
<p>7.	Python automatically executes the appropriate method based on the object.


```python
#Another Example: Polymorphism Without Inheritance
class SavingsAccount:
    def withdraw(self):
        print("Savings account withdrawal")

class CurrentAccount:
    def withdraw(self):
        print("Current account withdrawal")

class FixedDeposit:
    def withdraw(self):
        print("Fixed deposit withdrawal")
accounts = [SavingsAccount(), CurrentAccount(), FixedDeposit()]
for account in accounts:
    account.withdraw()
```

<p>Here, all classes have the same method name, even though they don't inherit from a common parent class. This is an example of duck typing in Python.
