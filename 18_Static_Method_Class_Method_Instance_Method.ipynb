<font color="blue" size="6">Static Methods </font>

<p>A Static Method is a method that belongs to a class but does not depend on a particular object or class.
<p>In Python, we use the @staticmethod decorator to create a static method.
<p>Simple definition: A static method is used to perform a task related to a class without accessing instance variables (self) or class variables (cls) automatically.


<font color="Blue"><b>1. Real-world bank application</font>
<p>Imagine a bank application.
<p>Before depositing or withdrawing money, the bank needs to validate the amount.
<p>For example:
<p>•	Deposit amount must be greater than zero.
<p>•	Withdrawal amount must be greater than zero.
<p>•	The amount should be valid.
    We can create a static method called validate_amount() to perform this validation.</p>


```python
class BankAccount:

    bank_name = "ABC Bank"

    def __init__(self, customer, balance):
        self.customer = customer
        self.balance = balance

    # Static Method
    @staticmethod
    def validate_amount(amount):
        if amount > 0:
            return True
        else:
            return False

    # Instance Method
    def deposit(self, amount):
        if BankAccount.validate_amount(amount):
            self.balance += amount
            print("Deposit Successful:", amount)
        else:
            print("Invalid Deposit Amount")

    def display_balance(self):
        print("Customer:", self.customer)
        print("Balance:", self.balance)


# Creating object
account = BankAccount("Arun", 10000)

# Calling static method
print(BankAccount.validate_amount(5000))

# Calling deposit method
account.deposit(5000)

# Display balance
account.display_balance()
```

<p>Explanation:
<p>1.	@staticmethod defines a static method.
<p>2.	validate_amount(amount) accepts an amount as a parameter.
<p>3.	It checks whether the amount is greater than zero.
<p>4.	If the amount is greater than zero, it returns True.
<p>5.	Otherwise, it returns False.
<p>6.	We can call the static method using the class name without creating an object.


<b>Calling Static Method in Two Ways
    <b><p>Method 1: Using class name</b>
<p>BankAccount.validate_amount(5000)
    <b><p>Method 2: Using object name</b>
<p>account = BankAccount("Arun", 10000)
<p>account.validate_amount(5000)

```python
#All Three Methods in One Bank Application
class BankAccount:

    bank_name = "ABC Bank"

    def __init__(self, customer, balance):
        self.customer = customer
        self.balance = balance

    # Instance Method
    def display_balance(self):
        print("Customer:", self.customer)
        print("Balance:", self.balance)

    # Class Method
    @classmethod
    def change_bank_name(cls, name):
        cls.bank_name = name

    # Static Method
    @staticmethod
    def validate_amount(amount):
        return amount > 0


account = BankAccount("Arun", 10000)

# Instance method
account.display_balance()

# Class method
BankAccount.change_bank_name("XYZ Bank")
print("Bank:", BankAccount.bank_name)

# Static method
print("Valid:", BankAccount.validate_amount(5000))
```
