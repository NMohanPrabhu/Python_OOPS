<Font color="Blue" size="6">Getters and Setters </font>

<p>Getter: A method used to retrieve or read the value of a private attribute.</p>
<p>Setter: A method used to modify or update the value of a private attribute.</p>


```python
class Student:
    def __init__(self, marks):
        self.marks = marks
s1 = Student(85)
print(s1.marks)
s1.marks = 95
print(s1.marks)
```

<p>Here, we can directly access and modify the attribute.
<p>But what happens if somebody assigns invalid marks?
<p>This is a problem because marks should be between 0 and 100.
<p>We can solve this using getters and setters.


```python
#Example using Getters and Setters
class Student:

    def __init__(self, marks):
        self.__marks = marks

    # Getter method
    def get_marks(self):
        return self.__marks

    # Setter method
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks!")

# Create object
s1 = Student(85)

# Access using getter
print(s1.get_marks())

# Modify using setter
s1.set_marks(95)

print(s1.get_marks())

# Assign invalid marks
s1.set_marks(150)
```

<b>Explanation 
<b><P>Statement	Purpose
<p>self.__marks = marks	Creates a private attribute
<p>get_marks()	Retrieves the marks
<p>set_marks(marks)	Updates the marks
<p>return self.__marks	Returns the attribute value
<p>if 0 <= marks <= 100	Validates the input



```python
#Practical Example: Bank Account

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative!")

account = BankAccount(5000)

print("Balance:", account.balance)

account.balance = 8000

print("Updated Balance:", account.balance)

account.balance = -2000

```

<p><b>What is @property.deleter?</b>
<p>Consider a bank account with a balance.

<p>We can:
<p>- Read the balance using a getter.
<p>- Update the balance using a setter.
<p>- Delete the balance property using a deleter.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    # Getter
    @property
    def balance(self):
        return self.__balance

    # Setter
    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid balance!")

    # Deleter
    @balance.deleter
    def balance(self):
        print("Deleting account balance")
        del self.__balance


# Creating object
account = BankAccount(5000)

# Getter
print("Current Balance:", account.balance)

# Setter
account.balance = 8000
print("Updated Balance:", account.balance)

# Deleter
del account.balance
```
