<font color="blue" size="6">Constructor</font>

<p>A constructor in Python is a special method that is automatically called when an object of a class is created. It is mainly used to initialize the object's data members (instance variables).
<p>In Python, the constructor is commonly written using the __init__() method.


```python
#Basic example of a constructor
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

p1 = Person("John", 25)
p1.display()
```

<font color="blue" size="4">Types of constructors</font>

```python
#Default-style constructor (no parameters)
class Student:
    def display(self):
        print("Welcome to Python")

s1 = Student()
s1.display()
```

```python
#We can also explicitly define a no-argument initializer:
class Student:
    def __init__(self):
        print("Student object created")

s1 = Student()
```

```python
#Parameterized constructor
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def display(self):
        print("Name:", self.name)
        print("Mark:", self.mark)

s1 = Student("Arun", 90)
s2 = Student("Priya", 95)

s1.display()
s2.display()
```

```python
#Real-world example: Bank account
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def display(self):
        print("Account holder:", self.name)
        print("Balance:", self.balance)

account1 = BankAccount("Ravi", 5000)

account1.deposit(1000)
account1.display()
```
