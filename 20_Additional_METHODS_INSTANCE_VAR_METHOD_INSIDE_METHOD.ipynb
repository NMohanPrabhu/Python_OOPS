<font color="blue" size="6">instance variables, instance methods, and methods inside methods</font>

```python
#1. Instance Variables
#Instance variables are variables that belong to an individual object.
#They are usually created using self inside the __init__() method.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Arun", 21)
s2 = Student("Bala", 22)

print(s1.name)
print(s2.name)
```

<p>Here:
<p>•	self.name is an instance variable.
<p>•	self.age is an instance variable.
<p>•	Each object has its own copy of these variables.


```python
#2. Instance Methods
#Instance methods are functions defined inside a class that operate on objects.
#They take self as their first parameter.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Arun", 21)

s1.display()
```

```python
#Here:
#•	display() is an instance method.
#•	self refers to the current object.
#•	s1.display() calls the method for object s1.
```

```python
#3. Method Inside Another Method
#In Python, one method can call another method belonging to the same class using self.
class Calculator:
    def add(self, a, b):
        return a + b
    def multiply(self, a, b):
        result = self.add(a, b)
        print("Addition result:", result)
        print("Multiplication:", a * b)
obj = Calculator()
obj.multiply(10, 20)
```

```python
#Here:
#•	add() is one method.
#•	multiply() is another method.
#•	self.add(a, b) calls the add() method from inside multiply().
#Important: This is called one method calling another method, not defining a nested function.
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def bonus(self):
        return self.salary * 0.10

    def display(self):
        print("Name:", self.name)
        print("Bonus:", self.bonus())

e1 = Employee("Arun", 30000)

e1.display()
```
