<font color="blue" size="6">Method Resolution Order</font>

<p>MRO stands for Method Resolution Order.
<p>MRO is the order in which Python searches for a method or attribute in classes when inheritance is involved.
<p>In simple words:
<p>When a child class calls a method, Python needs to decide which parent class's method should execute first.
<p>MRO determines that order.
<p>Python uses an algorithm called C3 Linearization to calculate MRO.


```python
#Simple banking example – Single inheritance
class BankAccount:
    def account_type(self):
        print("General Bank Account")
class SavingsAccount(BankAccount):
    def account_type(self):
        print("Savings Account")
obj = SavingsAccount()
obj.account_type()
```

<p>Python searches in this order:
<p>1.	SavingsAccount
<p>2.	BankAccount
<p>3.	object
<p>Because SavingsAccount has its own account_type() method, Python executes it immediately.
This is called method overriding.


```python
#MRO in multiple inheritance
class BankAccount:
    def display(self):
        print("Bank Account")


class SavingsAccount(BankAccount):
    def display(self):
        print("Savings Account")


class LoanAccount(BankAccount):
    def display(self):
        print("Loan Account")
class CustomerAccount(SavingsAccount, LoanAccount):
    pass
obj = CustomerAccount()
obj.display()
```

<p><b>Why does Python print Savings Account instead of Loan Account?</b>
<p><b>Because the class declaration is: class CustomerAccount(SavingsAccount, LoanAccount):</b>
<p>Python searches the classes in this MRO:
<p>MRO search order
<p>1.	CustomerAccount:First, check the child class.
<p>2.	SavingsAccount:The first parent class.
<p>3.	LoanAccount:The second parent class.
<p>4.	BankAccount:Common parent class.
<p>5.	object:Python's built-in base class.
<p>Therefore, SavingsAccount.display() executes first.


```python
#How to check MRO in Python?
#Python provides two ways to check MRO.
#Method 1: Using mro()
print(CustomerAccount.mro())
```

```python
#Method 2: Using __mro__
print(CustomerAccount.__mro__)
```

<p>MRO in hybrid inheritance (Diamond problem)
<p>This is one of the most important concepts in Python inheritance.
<p>Consider this banking structure:
<p>•	BankAccount is the parent.
<p>•	SavingsAccount inherits from BankAccount.
<p>•	LoanAccount inherits from BankAccount.
<p>•	CustomerAccount inherits from both SavingsAccount and LoanAccount.
<p>This creates a diamond-shaped inheritance structure.


```python
class BankAccount:
    def display(self):
        print("Bank Account")


class SavingsAccount(BankAccount):
    def display(self):
        print("Savings Account")


class LoanAccount(BankAccount):
    def display(self):
        print("Loan Account")


class CustomerAccount(SavingsAccount, LoanAccount):
    def display(self):
        print("Customer Account")


obj = CustomerAccount()
obj.display()
```

<p>Because Python first checks the child class itself.
<p>If you remove the display() method from CustomerAccount, Python checks SavingsAccount next.
<p>This is how MRO resolves the diamond problem without searching the common ancestor twice.


```python
#Understanding super() with MRO
#The super() function is closely related to MRO.
#It calls the next class in the MRO sequence, not necessarily the direct parent class.
#Let's understand this with a banking application.
class BankAccount:
    def display(self):
        print("Bank Account")


class SavingsAccount(BankAccount):
    def display(self):
        print("Savings Account")
        super().display()


class LoanAccount(BankAccount):
    def display(self):
        print("Loan Account")
        super().display()


class CustomerAccount(SavingsAccount, LoanAccount):
    def display(self):
        print("Customer Account")
        super().display()


obj = CustomerAccount()
obj.display()
```

<p>Step-by-step explanation:
  <p>  Step	Class executed	Reason
<p>1	CustomerAccount:	Initial method call
<p>2	SavingsAccount:	super() moves to next MRO class
<p>3	LoanAccount:	Next class in MRO
<p>4	BankAccount:	Next class in MRO


<p>When SavingsAccount calls super(), Python does not directly jump to BankAccount. 
<p>It goes to LoanAccount because that is the next class in the MRO.
<p>This is the most important point about super() in multiple inheritance.

<font color="blue" size="4"><b>How does Python calculate MRO? (C3 Linearization)</b></font>
<p>Python uses the C3 Linearization algorithm to determine a consistent method resolution order.
<p><b>The three important rules are:</b>
<p>1.	A child class must appear before its parent classes.
<p>2.	Parent classes must maintain the order specified in the class declaration.
<p>3.	A class must not appear more than once in the final MRO.


```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.mro())
```

<p>Here:
<p>•	D comes first.
<p>•	B comes before C because it was declared first.
<p>•	A appears only once because it is a common parent.
<p>•	object comes last.
