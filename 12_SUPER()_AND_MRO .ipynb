<font color="blue" size="6">SUPER() AND MRO (METHOD RESOLUTION ORDER)</font>

```python
#super() is used to call a method from the next class in the inheritance order.
#Example 1: Using super() with single inheritance
class Parent:
    def show(self):
        print("Parent class method")

class Child(Parent):
    def show(self):
        print("Child class method")
        super().show()

obj = Child()
obj.show()
```

```python
#•	obj.show() calls the Child's method.
#•	super().show() calls the next implementation of show() in the MRO, which is Parent's method.
```

```python
#Example 2: Using super() with constructors
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")
obj = Child()
```

```python
#What is MRO in Python?

#MRO stands for Method Resolution Order.

#It defines the order in which Python searches for a method or attribute in an inheritance hierarchy.

#Python uses the C3 Linearization algorithm to determine MRO.
```

```python
#MRO Example 1: Single inheritance
class A:
    pass

class B(A):
    pass

class C(B):
    pass

print(C.mro())
```

```python
#Example 2: Multiple inheritance
class A:
    def show(self):
        print("Class A")
class B(A):
    def show(self):
        print("Class B")
class C(A):
    def show(self):
        print("Class C")
class D(B, C):
    pass
obj = D()
obj.show()
print(D.mro())
```

```python
#Super working with MRO
class A:
    def show(self):
        print("A")
        super().show()
class B(A):
    def show(self):
        print("B")
        super().show()
class C(A):
    def show(self):
        print("C")
        super().show()
class D(B, C):
    def show(self):
        print("D")
        super().show()
obj = D()
obj.show()
```

```python
#The last super().show() in A raises an AttributeError because object does not have a show() method.
#The MRO is:
#D → B → C → A → object
#Notice something important:
#When super() is called inside B, it doesn't necessarily call B's parent A.
#Instead, it calls the next class in the MRO, which is C.
#This is why super() is particularly useful in multiple inheritance.
```
