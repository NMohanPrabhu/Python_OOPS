<font color="blue" size="6">Encapsulation</font>

```python
#example without encapsulation
class BankAccount:
    def __init__(self):
        self.balance = 10000

account = BankAccount()
print("Balance:", account.balance)
account.balance = -5000
print("Balance:", account.balance)
```

Problem:
The user can directly modify the balance and enter an invalid amount such as -5000.
We need a mechanism to protect the balance.
This is where encapsulation is useful.


```python
#Implementing Encapsulation using Private Variables
class BankAccount:

    def __init__(self):
        self.__balance = 10000

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance or invalid amount")

    def display_balance(self):
        print("Current Balance:", self.__balance)


account = BankAccount()
account.display_balance()
account.deposit(5000)
account.withdraw(2000)
account.display_balance()
```
