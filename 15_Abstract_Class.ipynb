<font color="blue" size="6">Abstract Base Class (ABC)</font>

<P><b>An Abstract Base Class (ABC) is a class that acts as a blueprint for other classes.</b>
<P><P>It defines methods that child classes must implement.
<P>In Python, we use:
<P>•	ABC from the abc module.
<P>•	@abstractmethod decorator to define abstract methods.
<P><b>Simple definition: An abstract class cannot be instantiated directly. It is designed to be inherited by child classes.</b>


```python
from abc import ABC, abstractmethod

# Abstract Base Class
class BankAccount(ABC):
    @abstractmethod
    def deposit(self, amount):
        pass
    @abstractmethod
    def withdraw(self, amount):
        pass
# Child class 1
class SavingsAccount(BankAccount):
    def deposit(self, amount):
        print("Savings Account Deposit:", amount)
    def withdraw(self, amount):
        print("Savings Account Withdrawal:", amount)
# Child class 2
class CurrentAccount(BankAccount):
    def deposit(self, amount):
        print("Current Account Deposit:", amount)
    def withdraw(self, amount):
        print("Current Account Withdrawal:", amount)

# Creating objects
savings = SavingsAccount()
current = CurrentAccount()

# Calling methods
savings.deposit(5000)
savings.withdraw(2000)

current.deposit(10000)
current.withdraw(3000)
```

<p>Step-by-step explanation:
<p>1.	from abc import ABC, abstractmethod imports the required components.
<p>2.	class BankAccount(ABC) creates an abstract base class.
<p>3.	@abstractmethod declares methods that must be implemented by concrete child classes.
<p>4.	pass indicates that the abstract method has no implementation in the parent class.
<p>5.	SavingsAccount and CurrentAccount inherit from BankAccount.
<p>6.	Both child classes implement the abstract methods.
<p>7.	We can now create objects of the child classes.


<p><b>What happens if we create an object of an abstract class</b></p>

```python
bank = BankAccount()
```

<p><b>Why?</b>
<p>Because an abstract class cannot be instantiated directly while it contains unimplemented abstract methods.


<b>What happens if a child class does not implement all abstract methods?

```python
class SavingsAccount(BankAccount):

    def deposit(self, amount):
        print("Deposit:", amount)


savings = SavingsAccount()
```

```python
#The child class must implement all inherited abstract methods before it can be instantiated.
```

<p>What is an Abstract Base Class in Python?
<p>An Abstract Base Class is a class that provides a common blueprint for child classes. It uses the ABC module and @abstractmethod decorator to enforce the implementation of required methods in subclasses.
<p>Remember: In our bank application, BankAccount defines what every account should do, while SavingsAccount and CurrentAccount define how those operations work.
