<font color="blue" size="6">Composition</font>

<p>Composition is an Object-Oriented Programming (OOP) concept in which one class contains an object of another class.
<p>In simple words:
<p>Composition means HAS-A relationship.
<p>For example:
<p>•	A Bank HAS-A Customer.
<p>•	A Bank Account HAS-A Transaction History.
<p>•	A Bank Account HAS-A Account Holder.


<p><b>Real-world example:Imagine a bank application.</b>
<p>Customer class stores customer details.
<p>BankAccount class stores account details.
<p>BankAccount contains a Customer object.
<p>Here, a BankAccount HAS-A Customer.


```python
# Customer class
class Customer:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    def display_customer(self):
        print("Customer Name:", self.name)
        print("Phone Number:", self.phone)
# BankAccount class
class BankAccount:
    def __init__(self, account_number, balance, customer):
        self.account_number = account_number
        self.balance = balance

        # Composition
        self.customer = customer
    def display_account(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)
        # Calling Customer class method
        self.customer.display_customer()
# Creating Customer object
customer1 = Customer("Arun", "9876543210")
# Passing Customer object to BankAccount
account1 = BankAccount(1001, 50000, customer1)
# Display account details
account1.display_account()
```

<p>Explanation:
<p>1.	We create a Customer class.
<p>2.	We create a BankAccount class.
<p>3.	Inside the BankAccount constructor, we receive a Customer object.
<p>4.	We store that object using self.customer = customer.
<p>5.	Now the BankAccount object can access Customer class methods and attributes.
<p>6.	This relationship is called Composition.


```python
#Another Example: Bank Account HAS-A Transaction
# Transaction class
class Transaction:
    def __init__(self):
        self.transactions = []
    def add_transaction(self, amount):
        self.transactions.append(amount)
    def display_transactions(self):
        print("Transaction History:", self.transactions)
# BankAccount class
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
        # Composition
        self.transaction = Transaction()
    def deposit(self, amount):
        self.balance += amount
        self.transaction.add_transaction(amount)
    def display(self):
        print("Balance:", self.balance)
        self.transaction.display_transactions()
# Creating BankAccount object
account = BankAccount(10000)
account.deposit(5000)
account.deposit(2000)
account.display()
```

<p>What is Composition in Python?
<p>Composition is an OOP concept where one class contains an object of another class to reuse its functionality. It represents a HAS-A relationship.
<p>Remember this simple example:
<p>Inheritance: SavingsAccount IS-A BankAccount.
<p>Composition: BankAccount HAS-A Customer.
