<font color="blue" size="6">Contructor Overloading Class Method</font>

In Python, we can use @classmethod to create alternative constructors. This allows us to create objects in different ways, depending on the data we have.

```python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    @classmethod
    def from_string(cls, data):
        name, mark = data.split(",")
        return cls(name, int(mark))

    def display(self):
        print("Name:", self.name)
        print("Mark:", self.mark)


# Method 1: Using the normal constructor
s1 = Student("Arun", 85)

# Method 2: Using the alternative constructor
s2 = Student.from_string("Priya,95")

s1.display()
s2.display()
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    @classmethod
    def create_account(cls, name):
        return cls(name, 0)

    def display(self):
        print("Account holder:", self.name)
        print("Balance:", self.balance)


# Normal constructor
a1 = BankAccount("Ravi", 5000)

# Alternative constructor
a2 = BankAccount.create_account("Priya")

a1.display()
a2.display()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    @classmethod
    def square(cls, side):
        return cls(side, side)

    def area(self):
        print(self.length * self.width)


r1 = Rectangle(10, 5)
r2 = Rectangle.square(4)

r1.area()
r2.area()
```
