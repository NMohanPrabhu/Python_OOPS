<p><b>What is a Decorator?</b>
<p>A Decorator in Python is a function that modifies or extends the behaviour of another function without changing its original source code.
<p>Decorators are used to add extra functionality to existing functions.


```python
#Simple Example of Decorator
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")

    return wrapper
@my_decorator
def greet():
    print("Hello Python")
greet()
```

<p><b>Explanation:</b>
<p>1.my_decorator() accepts another function as an argument.
<p>2.wrapper() adds extra functionality.
<p>3.func() calls the original function.
<p>4.@my_decorator applies the decorator to greet().
<p>5.When greet() is called, the wrapper function executes.


```python
#Decorator with Arguments
#A decorator can also work with functions that accept arguments.
#Example:
def my_decorator(func):
    def wrapper(a, b):
        print("Addition is performed")
        func(a, b)

    return wrapper

@my_decorator
def add(a, b):
    print(a + b)

add(10, 20)
```

```python
#Here, the wrapper function accepts two arguments and passes control to the original function.
```

```python
#Decorator with return Value
#A decorator can also return the result of the original function.
#Example:
def my_decorator(func):
    def wrapper(a, b):
        print("Calculating...")
        result = func(a, b)
        return result
    return wrapper
@my_decorator
def add(a, b):
    return a + b
print(add(10, 20))
```

```python
#6. Types of Decorators in Python
#A. Function Decorator:A function decorator modifies the behaviour of another function.
def decorator(func):
    def wrapper():
        print("Welcome")
        func()
    return wrapper
@decorator
def display():
    print("Python Programming")
display()
```

```python
#B. Class Decorator
#A class decorator is used to modify or enhance a class.
def my_decorator(cls):
    cls.message = "Hello Python"
    return cls
@my_decorator
class Student:
    pass
s = Student()
print(s.message)
```

```python
#C. Built-in Decorators
#Python provides some commonly used built-in decorators.
#1. @staticmethod:Used to define a method that does not require self or cls.
class Student:
    @staticmethod
    def display():
        print("Hello Student")
Student.display()
```

```python
#2. @classmethod:Used to define a method that receives the class as its first argument (cls).
class Student:
    name = "Arun"
    @classmethod
    def display(cls):
        print(cls.name)
Student.display()
```

```python
#3. @property:Used to access a method like an attribute.
class Student:
    def __init__(self, name):
        self.name = name
    @property
    def student_name(self):
        return self.name
s = Student("Arun")
print(s.student_name)
```

```python
#Decorator with *args and **kwargs
#*args and **kwargs allow a decorator to work with functions that accept different numbers of arguments.
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function started")
        result = func(*args, **kwargs)
        print("Function completed")
        return result
    return wrapper
@my_decorator
def add(a, b):
    return a + b
print(add(10, 20))
```

```python
#Real-Time Example: Execution Time Decorator
import time

def timer(func):
    def wrapper():
        start = time.time()

        func()

        end = time.time()

        print("Execution time:", end - start)

    return wrapper

@timer
def display():
    time.sleep(2)
    print("Hello Python")

display()
```

```python
#Here, the decorator calculates how long the function takes to execute.
```
